```python

class MacBook:
    def __init__(self, chip, ram):
        self.chip = chip
        self.ram = ram
        self.storage = "500gb"
        
    def __repr__(self):
        return f"최종 스펙: {self.chip}, {self.ram}, {self.storage}"
    
    def spec_custom(self, chip: str | None=None, ram: str | None=None):
        self.chip = chip
        self.ram = ram
    
    @classmethod
    def ordinary_book(cls):
        return cls(chip="m4", ram="8gb")
        
    @classmethod
    def good_book(cls):
        return cls(chip="m4pro", ram="48gb")

    
print(MacBook.ordinary_book())
print(MacBook.good_book())

class New(MacBook):
    def __init__(self, chip, ram):
        super().__init__(chip, ram)
        self.storage = "1tb"
        
    # # repr은 그대로 상속받으면 되므로 굳이 여기 새로 정의할 필요없으나...
    # # ...return은 상속이 안된다는 점을 기억하고자 남겼음
    # def __repr__(self):
    #     super().__repr__()
    #     return f""
        
    def spec_custom(self, chip, ram):
        super().spec_custom(chip, ram)
    
    @classmethod
    def my_only_book(cls):
        return cls(chip="m5max", ram="64gb")

basic_order = New("m5", "16gb")
print(basic_order)

print(New.my_only_book())
```

```python
x = 1

def first(arg1):
    print(x)
    
    # second 정의
    def second(arg2):
        y = 2
        
        # third 정의
        def third(arg3):
            global x
            nonlocal y
            z = 3            
            # 변수 이름 혼동을 피하기 위해 매개변수명을 명확히 구분했습니다.
            return x + y + z + arg3
            
        # third를 실행하여 결과(숫자)를 반환 (arg2를 전달)
        return third(arg2)
    
    # second를 실행하여 결과(숫자)를 반환 (arg1을 전달)
    return second(arg1)

# first(3)이 실행되면:
# 1. second(3) 호출 -> 2. third(3) 호출 -> 3. 1+2+3+3 = 9 계산
final_result = first(3) 

print(final_result)
# func(4)는 불가능합니다. final_result는 이미 숫자 9이기 때문입니다.
```

```python
target1 = "아많다많다많다많아"
target2 = "아정말많다많다많아"

check_False = []

def is_palindrome(text):
  for c in range(int(len(text)/2)):
    if text[c] == text[-1-c]:
        check_False.append(bool(text[c] == text[-1-c]))
    else:
        check_False.append(bool(text[c] == text[-1-c]))

  if False in check_False:
    return print("회문이 아닙니다.")
  else:
    return print("회문이 맞습니다.")
```

```python
targets = ['기러기', '까마귀', '토마토', '스윗소스', '스위스', '인도인', '한국인', '별똥별', '우영우', '우향우']
classified_dict = {"회문O": [], "회문X": []}
    
for t in targets:
    check_False = []
    for c in range(int(len(t)/2)):
        if t[c] == t[-1-c]:
            check_False.append(bool(t[c] == t[-1-c]))
        else:
            check_False.append(bool(t[c] == t[-1-c]))
            
    if False in check_False:
        classified_dict["회문X"].append(t)
    else:
        classified_dict["회문O"].append(t)

print(classified_dict)

true_count = len(classified_dict["회문O"])
false_count = len(classified_dict["회문X"])

print(true_count)
print(true_count/(true_count+false_count))
```

```python
class macbook:
    def __init__(self, chip, ram):
        self.chip = "m5"
        self.ram = "8gb"
        self.storage = "1tb"
        
    
    def spec(self, chip, ram):
        self.chip = chip
        self.ram = ram
        
        return f"a normal macbook wearing {self.chip}, {self.ram}, {self.storage}"
        
    @classmethod
    def my_only_book(cls):
        return f"this is my the very book wearing {cls(chip="m5max", ram="64gb")}"
        
order = macbook()
print(order.spec("m4pro", "16gb"))
print(macbook.my_only_book())
```
